ConversationBufferMemory from langchain_classic.agents import
AgentExecutor, ConversationalChatAgent from langchain_community.tools

WikipediaAPIWrapper import os import sys

import warnings from langchain_core.\_api.deprecation import
LangChainDeprecationWarning

# Initialize the LLM

llm = ChatGroq( model=“llama-3.1-8b-instant”, temperature=0 )

print(“Groq LLM initialized”)

# Initialize Conversation Memory

memory = ConversationBufferMemory( memory_key=“chat_history”,
return_messages=True )

print(“Memory initialized”)

# Set Up External Knowledge Tool (Wikipedia)

wiki_tool = WikipediaQueryRun(
api_wrapper=WikipediaAPIWrapper(top_k\_results=2) )

tools = \[wiki_tool\]

print(“Wikipedia tool ready”)

# Create Conversational Agent

agent = ConversationalChatAgent.from_llm_and_tools( llm=llm,
tools=tools, system_message=“You are a helpful conversational knowledge
assistant.” )

# Create Agent Executor

agent_executor = AgentExecutor( agent=agent, tools=tools, memory=memory,
verbose=False  
)

print(“Agent initialized successfully”)

# Command-Line Chat Loop

print(“Knowledge Bot”) print(“Type ‘exit’ or ‘quit’ to stop”)

while True: user_input = input(“You:”)

    if user_input.lower() in ["exit", "quit"]:
        print("Thank you for your time")
        break

    sys.stdout.write("Bot: Please wait...")
    sys.stdout.flush()

    response = agent_executor.invoke({"input": user_input})

    print("\rBot:", response["output"], " " * 10, "\n")
